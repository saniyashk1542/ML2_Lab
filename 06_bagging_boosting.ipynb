```python
# ============================================================
# BAGGING AND BOOSTING - STUDENT PLACEMENT PREDICTION
# ============================================================

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
```

```python

# ============================================================
# 1. CREATE SUPERVISED DATASET
# ============================================================

data = [
    [8.7, 3, 2, 5, 82, 85, "Yes"],
    [7.2, 1, 1, 2, 65, 70, "No"],
    [8.1, 2, 2, 4, 78, 80, "Yes"],
    [6.8, 0, 1, 1, 55, 60, "No"],
    [9.0, 3, 3, 6, 90, 92, "Yes"],
    [7.5, 1, 2, 3, 68, 72, "No"],
    [8.5, 2, 3, 5, 85, 88, "Yes"],
    [6.5, 0, 0, 1, 50, 55, "No"],
    [8.9, 3, 2, 6, 88, 90, "Yes"],
    [7.0, 1, 1, 2, 60, 65, "No"],
    [8.3, 2, 2, 5, 80, 84, "Yes"],
    [6.9, 0, 1, 2, 58, 62, "No"],
    [9.1, 3, 3, 7, 92, 95, "Yes"],
    [7.4, 1, 1, 3, 66, 69, "No"],
    [8.6, 2, 3, 5, 86, 89, "Yes"],
    [6.7, 0, 0, 1, 52, 58, "No"],
    [8.0, 2, 1, 4, 75, 79, "Yes"],
    [7.1, 1, 1, 2, 63, 67, "No"],
    [8.8, 3, 3, 6, 89, 91, "Yes"],
    [6.6, 0, 1, 1, 54, 59, "No"],
    [8.4, 2, 2, 5, 81, 86, "Yes"],
    [7.3, 1, 2, 3, 67, 71, "No"],
    [9.2, 3, 3, 7, 94, 96, "Yes"],
    [6.4, 0, 0, 1, 48, 53, "No"],
    [8.2, 2, 2, 4, 79, 83, "Yes"],
    [7.6, 1, 1, 3, 69, 73, "No"],
    [8.9, 3, 2, 6, 87, 91, "Yes"],
    [6.8, 0, 1, 2, 57, 61, "No"],
    [8.5, 2, 3, 5, 84, 87, "Yes"],
    [7.0, 1, 1, 2, 61, 66, "No"]
]

columns = [
    "CGPA",
    "Internships",
    "Certifications",
    "Projects",
    "AptitudeScore",
    "CommunicationScore",
    "Placed"
]

df = pd.DataFrame(data, columns=columns)

```

```python

# ============================================================
# 2. DISPLAY DATASET
# ============================================================

print("=" * 70)
print("STUDENT PLACEMENT DATASET")
print("=" * 70)

print(df.to_string(index=False))

```

```python

# ============================================================
# 3. PREPARE INPUT AND TARGET
# ============================================================

X = df.drop("Placed", axis=1)

y = df["Placed"]

# Convert Yes/No into 1/0
encoder = LabelEncoder()
y = encoder.fit_transform(y)

```

```python

# ============================================================
# 4. SPLIT DATASET
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python

# ============================================================
# 5. BAGGING MODEL
# ============================================================

base_model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

bagging_model = BaggingClassifier(
    estimator=base_model,
    n_estimators=50,
    random_state=42
)

bagging_model.fit(
    X_train,
    y_train
)

bagging_prediction = bagging_model.predict(X_test)

```

```python

# ============================================================
# 6. BOOSTING MODEL
# ============================================================

boosting_model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(
        max_depth=2,
        random_state=42
    ),
    n_estimators=50,
    learning_rate=0.8,
    random_state=42
)

boosting_model.fit(
    X_train,
    y_train
)

boosting_prediction = boosting_model.predict(X_test)

```

```python

# ============================================================
# 7. PERFORMANCE FUNCTION
# ============================================================

def evaluate_model(name, y_true, y_pred):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")

    print("\nClassification Report:")

    print(
        classification_report(
            y_true,
            y_pred,
            target_names=["Not Placed", "Placed"],
            zero_division=0
        )
    )

    return [
        accuracy,
        precision,
        recall,
        f1
    ]

```

```python

# ============================================================
# 8. EVALUATE BAGGING
# ============================================================

bagging_scores = evaluate_model(
    "BAGGING CLASSIFIER",
    y_test,
    bagging_prediction
)

```

```python

# ============================================================
# 9. EVALUATE BOOSTING
# ============================================================

boosting_scores = evaluate_model(
    "BOOSTING - ADABOOST",
    y_test,
    boosting_prediction
)

```

```python

# ============================================================
# 10. COMPARE MODELS
# ============================================================

comparison = pd.DataFrame(
    [
        bagging_scores,
        boosting_scores
    ],
    columns=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],
    index=[
        "Bagging",
        "Boosting"
    ]
)

print("\n")
print("=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

print(comparison.round(4))

```

```python

# ============================================================
# 11. FIND BEST MODEL
# ============================================================

best_model = comparison["Accuracy"].idxmax()

print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    f"Based on Accuracy, the better model is: {best_model}"
)
```
