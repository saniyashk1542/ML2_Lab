```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/kaggle/input/datasets/dhrubangtalukdar/indian-premier-league-ipl-all-matches2008-2025/ipl.csv")
```

```python
matches_per_season = df["season"].value_counts().sort_index()

matches_per_season
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=matches_per_season.index,
    y=matches_per_season.values
)

plt.title("Number of IPL Matches by Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")

plt.show()
```

```python
team_wins = df["winner"].value_counts()

team_wins
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=team_wins.values,
    y=team_wins.index
)

plt.title("Number of IPL Match Wins by Team")
plt.xlabel("Number of Wins")
plt.ylabel("Team")

plt.show()
```

```python
pom = df["player_of_match"].value_counts().head(10)

pom
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=pom.values,
    y=pom.index
)

plt.title("Top 10 Players by Player of the Match Awards")
plt.xlabel("Number of Awards")
plt.ylabel("Player")

plt.show()
```

```python
toss_decision = df["toss_decision"].value_counts()

toss_decision
```

```python
plt.figure(figsize=(6, 5))

sns.barplot(
    x=toss_decision.index,
    y=toss_decision.values
)

plt.title("IPL Toss Decisions")
plt.xlabel("Toss Decision")
plt.ylabel("Number of Matches")

plt.show()
```

```python
toss_match_winner = df["toss_winner"] == df["winner"]

toss_match_winner.value_counts()
```

```python
toss_win_percentage = toss_match_winner.mean() * 100

print(f"Toss winner also won the match: {toss_win_percentage:.2f}%")
```

```python
top_venues = df["venue"].value_counts().head(10)

top_venues
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_venues.values,
    y=top_venues.index
)

plt.title("Top 10 IPL Venues by Number of Matches")
plt.xlabel("Number of Matches")
plt.ylabel("Venue")

plt.show()
```
